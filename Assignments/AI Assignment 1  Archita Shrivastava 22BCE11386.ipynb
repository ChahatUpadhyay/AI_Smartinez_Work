```python
import pandas as pd

```

```python



file_path = "House Price India.csv"  # Update with your file path if needed
df = pd.read_csv(file_path)

```

```python


df.info()
df.head()
```

```python

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


```

```python
print(df.describe())

df.hist(figsize=(15, 12), bins=30)
plt.tight_layout()
plt.show()

# Countplot for categorical/binary columns
categorical_cols = ['waterfront present', 'number of bedrooms', 'number of floors', 'condition of the house']
for col in categorical_cols:
    sns.countplot(x=col, data=df)
    plt.title(f'Distribution of {col}')
    plt.show()
```

```python
# Price vs Living Area
sns.scatterplot(x='living area', y='Price', data=df)
plt.title('Price vs Living Area')
plt.show()

# Price vs Number of Bedrooms
sns.boxplot(x='number of bedrooms', y='Price', data=df)
plt.title('Price vs Number of Bedrooms')
plt.show()
```

```python

sns.boxplot(x='waterfront present', y='Price', data=df)
plt.title('Waterfront Presence vs Price')
plt.show()

# Grade vs Price
sns.boxplot(x='grade of the house', y='Price', data=df)
plt.title('Grade vs Price')
plt.xticks(rotation=90)
plt.show()
```

```python
plt.figure(figsize=(15, 12))
corr_matrix = df.corr()
sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```
