```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import kagglehub

# Download latest version
path = "/content/House Price India.csv"
print("Path to dataset file:", path)

# Load the dataset
df = pd.read_csv(path)
```

```python
# Display basic info
display(df.head())
display(df.info())
display(df.describe())
```

```python
# Univariate Analysis
plt.figure(figsize=(12, 6))
df.hist(bins=30, figsize=(12, 10), color='skyblue', edgecolor='black')
plt.suptitle("Univariate Analysis - Histogram", fontsize=14)
plt.show()
```

```python
# Bi-Variate Analysis (Scatterplot for numerical columns)
if len(df.select_dtypes(include=['number']).columns) >= 2:
    plt.figure(figsize=(8, 6))
    sns.scatterplot(x=df.select_dtypes(include=['number']).columns[0],
                    y=df.select_dtypes(include=['number']).columns[1],
                    data=df, hue=df.columns[-1], palette='viridis')
    plt.title("Bivariate Analysis - Scatterplot")
    plt.show()
```

```python
# Multi-Variate Analysis (Pairplot for numerical features)
sns.pairplot(df.select_dtypes(include=['number']))
plt.show()
```

```python
# Perform descriptive statistics
display(df.describe())

# Handle Missing Values
missing_values = df.isnull().sum()
print("Missing Values in Each Column:")
print(missing_values)
```

```python
#Fill missing values with median (numerical columns) and mode (categorical columns)
df.fillna({col: df[col].median() if df[col].dtype != 'O' else df[col].mode()[0]
          for col in df.columns}, inplace=True)

print("\nMissing values after handling:")
print(df.isnull().sum())
```
